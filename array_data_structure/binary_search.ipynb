```python
# Impleemntation of Binary Search
# Function definition
# Method 1: Recursion
from typing import Any


def binarySearch(arr: list, i: int, j: int, x: Any):
    while i < j:
        mid = i + (j-i)//2
        if arr[mid] == x:
            return mid
        
        elif x > arr[mid]:
            return binarySearch(arr, mid+1, j, x)
        
        else:
            return binarySearch(arr, i, mid-1, x)
        
    return -1



# Driver Code
# Sorted Array
arr = [2, 5, 10, 14, 18, 22, 27, 35, 40, 59]
x = 40
i = 0
j = len(arr) - 1

# Function Call
result = binarySearch(arr, i, j, x)
print("searched element is present at index", result)
```
