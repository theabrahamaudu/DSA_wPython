```python
# Reverse Linked List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def reverseList(head: ListNode) -> ListNode:
    prev: ListNode = None
    curr: ListNode = head

    while curr is not None:
        temp: ListNode = curr.next
        curr.next = prev
        prev = curr
        curr = temp

    return prev

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
getList(reverseList(head))
```

```python
# Middle of the Linked List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def middleNode(head: ListNode) -> ListNode:
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next 

    return slow

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
getList(middleNode(head))
```

```python
# Linked List Cycle
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def hasCycle(head: ListNode):
    if head is None:
        return False

    slow = head
    fast = head.next

    while slow != None or fast != None:
        if fast == None or fast.next == None:
            return False

        if fast == slow:
            return True

        slow = slow.next
        fast = fast.next.next

    return False

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node2

head = node1
hasCycle(head)
```

```python
# Merge Two Sorted Lists
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def mergeTwoLists(list1: ListNode, list2: ListNode) -> ListNode:
    dummy = ListNode(1)
    merge = dummy

    while list1 != None and list2 != None:
        if list1.data <= list2.data:
            merge.next = list1
            list1 = list1.next
        else:
            merge.next = list2
            list2 = list2.next

        merge = merge.next

    merge.next = list1 if list1 else list2

    return dummy.next

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)
node5 = ListNode(0)

node1.next = node2
node2.next = node3
node3.next = node4


list1 = node1
list2 = node5

getList(mergeTwoLists(list1, list2))
```

```python
# Remove the Nth Node from End of List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def removeNthFromEnd(head: ListNode, n: int) -> ListNode:
    dummy = ListNode(1)
    dummy.next = head

    front = dummy
    back = dummy

    for _ in range(n):
        front = front.next

    while front != None:
        front = front.next
        back = back.next

    back.next = back.next.next

    return dummy.next

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1

getList(removeNthFromEnd(head, 2))
```

```python
# Reorder List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def reorderList(head: ListNode):
    if head == None:
        return

    slow = head
    fast = head

    while fast != None and fast.next != None:
        slow = slow.next
        fast = fast.next.next

    curr = slow
    prev: ListNode = None
    temp: ListNode = None 

    while curr != None:
        temp = curr.next
        curr.next = prev
        prev = curr
        curr = temp

    first = head
    second = prev

    while second.next != None:
        temp = first.next
        first.next = second
        first = temp

        temp = second.next
        second.next = first
        second = temp

    return head

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

getList(reorderList(node1))
```

```python
# Find the Duplicate Number
def findDuplicate(nums: list[int]) -> int:
    slow = nums[0]
    fast = nums[nums[0]]

    while slow != fast:
        slow = nums[slow]
        fast = nums[nums[fast]]

    slow = 0
    while slow != fast:
        slow = nums[slow]
        fast = nums[fast]

    return slow

nums = [1, 2, 3, 2, 4, 5]     
findDuplicate(nums)
```

```python
# Copy List with Random Pointer
class Node:
    def __init__(self, val: int) -> None:
        self.val = val
        self.next: Node = None
        self.random: Node = None

class Solution:
    visitedNode: dict[Node, Node] = dict()

    def copyRandomList(self, head: Node) -> Node:
        if head == None:
            return None

        if head in self.visitedNode.keys():
            return self.visitedNode[head]

        node = Node(head.val)

        self.visitedNode[head] = node
        node.next = self.copyRandomList(head.next)
        node.random = self.copyRandomList(head.random)

        return node

def getList(head: Node) -> list:
    vals = []
    while head:
        vals.append([
                head.val,
                head.next.val if head.next else head.next, 
                head.random.val if head.random else head.random
            ])
        head = head.next
    return vals

node1 = Node(4)
node2 = Node(2)
node3 = Node(1)
node4 = Node(3)

node1.next = node2
node2.next = node3
node3.next = node4

node1.random = node3
node2.random = node1
node3.random = None
node4.random = node2

getList(Solution().copyRandomList(node1))

    
```

```python
# Merge K Sorted Lists
from queue import PriorityQueue

class ListNode:
    def __init__(self, val) -> None:
        self.val = val
        self.next: ListNode = None

def mergeKLists(lists: list[ListNode]):
    minHeap = PriorityQueue()
    for lst in lists:
        while lst != None:
            minHeap.put(lst.val)
            lst = lst.next
    dummy: ListNode = ListNode(1)
    merge = dummy

    while minHeap.qsize() > 0:
        merge.next = ListNode(minHeap.get())
        merge = merge.next

    return dummy.next

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.val)
        head = head.next
    return vals

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node3.next = node4

getList(mergeKLists([node1, node3]))
```

```python
# Reverse Nodes in K-Group
class ListNode:
    def __init__(self, val) -> None:
        self.val = val
        self.next: ListNode = None

def reverseLinkedList(head: ListNode, k: int):
    newHead: ListNode = None
    ptr: ListNode = head

    while k > 0:
        nextNode = ptr.next
        ptr.next = newHead
        newHead = ptr
        ptr = nextNode

        k -= 1

    return newHead

def reverseKGroup(head: ListNode, k: int):
    ptr: ListNode = head
    ktail: ListNode = None
    newHead: ListNode = None

    while ptr != None:
        count = 0
        ptr = head

        while count < k and ptr != None:
            ptr = ptr.next
            count += 1

        if count == k:
            revHead: ListNode = reverseLinkedList(head, k)

            if newHead == None:
                newHead = revHead

            if ktail != None:
                ktail.next = revHead

            ktail = head
            head = ptr

    if ktail != None:
        ktail.next = head

    return head if newHead == None else newHead

def getList(head: ListNode):
    itms = []

    while head != None:
        itms.append(head.val)
        head = head.next

    return itms

    
node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

k = 2

getList(reverseKGroup(node1, k))
```

```python
# Flatten Binary Tree to Linked List
class TreeNode:
    def __init__(self, val: int) -> None:
        self.val = val
        self.left: TreeNode = None
        self.right: TreeNode = None

def flattenTree(node: TreeNode) -> TreeNode:
    if node == None:
        return None

    if node.left == None and node.right == None:
        return node

    leftTail = flattenTree(node.left)
    rightTail = flattenTree(node.right)

    if leftTail != None:
        leftTail.right = node.right
        node.right = node.left
        node.left = None

    return leftTail if rightTail is None else rightTail

def getList(node: TreeNode):
    itms = []

    while node:
        itms.append(node.val)
        itms.append(node.left.val if node.left else None)
        node = node.right

    return itms

node1 = TreeNode(4)
node2 = TreeNode(2)
node3 = TreeNode(1)
node4 = TreeNode(3)
node5 = TreeNode(6)
node6 = TreeNode(4)

node1.left = node2
node1.right = node3
node2.left = node4
node2.right = node6
node3.right = node5

flattenTree(node1)
getList(node1)
    
```
