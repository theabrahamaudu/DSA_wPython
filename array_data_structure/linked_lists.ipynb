```python
# Reverse Linked List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def reverseList(head: ListNode) -> ListNode:
    prev: ListNode = None
    curr: ListNode = head

    while curr is not None:
        temp: ListNode = curr.next
        curr.next = prev
        prev = curr
        curr = temp

    return prev

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
getList(reverseList(head))
```
