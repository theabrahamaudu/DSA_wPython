```python
# Reverse Linked List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def reverseList(head: ListNode) -> ListNode:
    prev: ListNode = None
    curr: ListNode = head

    while curr is not None:
        temp: ListNode = curr.next
        curr.next = prev
        prev = curr
        curr = temp

    return prev

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
getList(reverseList(head))
```

```python
# Middle of the Linked List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def middleNode(head: ListNode) -> ListNode:
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next 

    return slow

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
getList(middleNode(head))
```

```python
# Linked List Cycle
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def hasCycle(head: ListNode):
    if head is None:
        return False

    slow = head
    fast = head.next

    while slow != None or fast != None:
        if fast == None or fast.next == None:
            return False

        if fast == slow:
            return True

        slow = slow.next
        fast = fast.next.next

    return False

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node2

head = node1
hasCycle(head)
```

```python
# Merge Two Sorted Lists
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def mergeTwoLists(list1: ListNode, list2: ListNode) -> ListNode:
    dummy = ListNode(1)
    merge = dummy

    while list1 != None and list2 != None:
        if list1.data <= list2.data:
            merge.next = list1
            list1 = list1.next
        else:
            merge.next = list2
            list2 = list2.next

        merge = merge.next

    merge.next = list1 if list1 else list2

    return dummy.next

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)
node5 = ListNode(0)

node1.next = node2
node2.next = node3
node3.next = node4


list1 = node1
list2 = node5

getList(mergeTwoLists(list1, list2))
```

```python
# Remove the Nth Node from End of List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def removeNthFromEnd(head: ListNode, n: int) -> ListNode:
    dummy = ListNode(1)
    dummy.next = head

    front = dummy
    back = dummy

    for _ in range(n):
        front = front.next

    while front != None:
        front = front.next
        back = back.next

    back.next = back.next.next

    return dummy.next

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1

getList(removeNthFromEnd(head, 2))
```

```python
# Add Two Numbers
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def addTwoNumbers(l1: ListNode, l2: ListNode) -> ListNode:
    dummy = ListNode(0)
    ans = dummy
    carry: int = 0

    while l1 is not None or l2 is not None or carry != 0:
        x = l1.data if l1 != None else 0
        y = l2.data if l2 != None else 0
        total = carry + x + y
        carry = total//10
        ans.next = ListNode(total%10)
        ans = ans.next

        if l1 != None:
            l1 = l1.next
        if l2 != None:
            l2 = l2.next

    return dummy.next

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node3.next = node4

getList(addTwoNumbers(node1, node3))
```

```python
# Reorder List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals

def reorderList(head: ListNode):
    if head == None:
        return

    slow = head
    fast = head

    while fast != None and fast.next != None:
        slow = slow.next
        fast = fast.next.next

    curr = slow
    prev: ListNode = None
    temp: ListNode = None 

    while curr != None:
        temp = curr.next
        curr.next = prev
        prev = curr
        curr = temp

    first = head
    second = prev

    while second.next != None:
        temp = first.next
        first.next = second
        first = temp

        temp = second.next
        second.next = first
        second = temp

    return head

node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

getList(reorderList(node1))
```
