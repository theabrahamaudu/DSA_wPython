```python
# search for an element "15" and if its not present in an array, return the index of that
# element
# if the element is not present, return -1

# Time complexity: O(n)
# Space complexity: O(1) 
# Function definition
def LinearSearch(x, arr):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1

# Driver code
arr = [2, 1, 8, 9, 12, 15, 11, 19]
x = 20

# Function Definition
result = LinearSearch(x, arr)
print("Searching element is present at index", result)
```

```python
## Best time to buy and sell stock
# function definition
def findMaxProfit(prices):
    min_price = float('inf')
    max_profit = 0

    for i in range(len(prices)):
        if prices[i] < min_price:
            min_price = prices[i]
        elif prices[i] - min_price > max_profit:
            max_profit = prices[i] - min_price
    return max_profit



# Driver code
prices = [7, 5, 3, 6, 4, 15]
maxProfit = findMaxProfit(prices)
print("The maximum possible profit is", maxProfit)
```

```python
## Collinearity

# Aprroach 1: Using slope concept
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    if ((y2 - y1) * (x3 - x2) == (y3 - y2)*(x2 - x1)):
        return True
    return False


# Driver code
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)

```

```python
# Approach 2: Using area of triangle
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    area = 0.5 * (x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))

    if area == 0:
        return True
    return False

# Driver code
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)
```

```python
## Majority Element - Boyer-Moore Voting Algorithm

# Function definition
def majorityElement(nums):
    # find candidate
    count = 0
    candidate = None
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)

    # validate majority status
    validation_count = 0
    for num in nums:
        if num == candidate:
            validation_count += 1
    
    if validation_count > len(nums)//2:
        return candidate
    return None
    


# Driver Code
# nums = [2, 2, 1, 1, 1, 2, 2]
nums = [2, 2, 7, 3, 4]
result = majorityElement(nums)
print("The majority element is", result)
```

```python
## Sort Colours -- Two-Pointer
# function definition
def sortColours(nums):
    curr = 0
    p1 = 0
    p2 = len(nums) - 1
    while curr < p2:
        if nums[curr] == 0:
            nums[curr], nums[p1] = nums[p1], nums[curr]
            p1 += 1
            curr += 1
        elif nums[curr] == 2:
            nums[curr], nums[p2] = nums[p2], nums[curr]
            p2 -= 1
            curr += 1
        else:
            curr += 1
    return nums

# Driver Code
nums = [2, 0, 2, 1, 1, 0]
result = sortColours(nums)
print(result)
```
