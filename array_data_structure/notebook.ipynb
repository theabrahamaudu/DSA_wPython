```python
# search for an element "15" and if its not present in an array, return the index of that
# element
# if the element is not present, return -1

# Time complexity: O(n)
# Space complexity: O(1) 
# Function definition
def LinearSearch(x, arr):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1

# Driver code
arr = [2, 1, 8, 9, 12, 15, 11, 19]
x = 20

# Function Definition
result = LinearSearch(x, arr)
print("Searching element is present at index", result)
```

```python
## Best time to buy and sell stock
# function definition
def findMaxProfit(prices):
    min_price = float('inf')
    max_profit = 0

    for i in range(len(prices)):
        if prices[i] < min_price:
            min_price = prices[i]
        elif prices[i] - min_price > max_profit:
            max_profit = prices[i] - min_price
    return max_profit



# Driver code
prices = [7, 5, 3, 6, 4, 15]
maxProfit = findMaxProfit(prices)
print("The maximum possible profit is", maxProfit)
```

```python
## Collinearity

# Aprroach 1: Using slope concept
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    if ((y2 - y1) * (x3 - x2) == (y3 - y2)*(x2 - x1)):
        return True
    return False


# Driver code
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)

```

```python
# Approach 2: Using area of triangle
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    area = 0.5 * (x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))

    if area == 0:
        return True
    return False

# Driver code
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)
```

```python
## Majority Element - Boyer-Moore Voting Algorithm

# Function definition
def majorityElement(nums):
    # find candidate
    count = 0
    candidate = None
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)

    # validate majority status
    validation_count = 0
    for num in nums:
        if num == candidate:
            validation_count += 1
    
    if validation_count > len(nums)//2:
        return candidate
    return None
    


# Driver Code
# nums = [2, 2, 1, 1, 1, 2, 2]
nums = [2, 2, 7, 3, 4]
result = majorityElement(nums)
print("The majority element is", result)
```

```python
## Sort Colours -- Two-Pointer
# function definition
def sortColours(nums):
    curr = 0
    p1 = 0
    p2 = len(nums) - 1
    while curr < p2:
        if nums[curr] == 0:
            nums[curr], nums[p1] = nums[p1], nums[curr]
            p1 += 1
            curr += 1
        elif nums[curr] == 2:
            nums[curr], nums[p2] = nums[p2], nums[curr]
            p2 -= 1
            curr += 1
        else:
            curr += 1
    return nums

# Driver Code
nums = [2, 0, 2, 1, 1, 0]
result = sortColours(nums)
print(result)
```

```python
## Two Sum Problem

def twoSum(arr, target):
    map = dict()

    for i in range(len(arr)):
        complement = target - arr[i]

        if complement in map.keys():
            return [map[complement], i]
        map[arr[i]] = i
    return []



# Driver Code
arr = [1, 2, 7, 11]
target = 13
print(twoSum(arr, target))
```

```python
## Contains duplicates
def hasDuplicates(arr):
    tracker = set()

    for num in arr:
        if num in tracker:
            return True
        tracker.add(num)
    return False

# Driver code
arr = [1, 3, 4, 6, 1 ,1, 2, 10]
print(hasDuplicates(arr))
```

```python
## Contains Duplicates II
def containsNearbyDuplicate(arr, k):
    tracker = set()
    for i in range(len(arr)):
        if arr[i] in tracker:
            return True
        tracker.add(arr[i])
        if i >= k:
            tracker.remove(arr[i-k])
    return False


# Driver Code 
arr = [1, 3, 4, 1, 6, 7, 1 , 2, 10]
k = 3
print(containsNearbyDuplicate(arr, k))
```

```python
## Anagrams
from array import array
def isAnagram(s: str, t: str):
    s = s.lower()
    t = t.lower()
    if len(s) != len(t):
        return False
    charCounts = array('i', [0] * 26)
    
    for i in range(len(s)):
        charCounts[ord(s[i]) - ord('a')]+=1
        charCounts[ord(t[i]) - ord('a')]-=1
    
    for count in charCounts:
        if count != 0:
            return False
    return True
        

# Driver code
s = "chars"
t = "schar"

print(isAnagram(s, t))
```

```python
## Group Anagrams
def groupAnagrams(strs: list[str]):
    if len(strs) == 0:
        return []
    
    ansMap = dict()
    
    for s in strs:
        counts = array('i', [0] * 26)
        for i in range(len(s)):
            counts[ord(s[i]) - ord('a')]+=1
        if str(counts) in ansMap.keys():
            ansMap[str(counts)].append(s)
        else:
            ansMap[str(counts)] = [s]
    return [value for value in ansMap.values()]

# Driver code
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(groupAnagrams(strs))
```

```python
## Top K Frequent Elements
import heapq
def topKFrequent(nums, k):
    if len(nums) == k:
        return nums
    
    count = dict()
    min_heap = []
    for num in nums:
        if num in count:
            count[num] += 1
        else:
            count[num] = 1
    for key, value in count.items():
        heapq.heappush(min_heap, (value, key))
        if len(min_heap) > k:
            heapq.heappop(min_heap)

    return [num for _, num in min_heap]

# Driver Code
nums = [1,1,1,1,3,3,2,2,2,26,6,6,6,6,6,]
k = 2
print(topKFrequent(nums,k))
    
```

```python
## Roman to Integer

values = {
    "I": 1,
    "V": 5,
    "X": 10,
    "L": 50,
    "C": 100,
    "D": 500,
    "M": 1000,
    "IV": 4,
    "IX": 9,
    "XL": 40,
    "XC": 90,
    "CD": 400,
    "CM": 900
}

def romanToInt(s):
    sum = 0
    i = 0
    while i < len(s):
        if i < len(s)-1:
            twoSymbols = s[i:i+2]
            if twoSymbols in values:
                sum += values[twoSymbols]
                i+=2
                continue
            oneSymbol = s[i:i+1]
            sum += values[oneSymbol]
            i+=1
    return sum

# Driver code
s = 'CMIV'
result = romanToInt(s)
print(result)
```
