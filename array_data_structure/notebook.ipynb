```python
# search for an element "15" and if its not present in an array, return the index of that
# element
# if the element is not present, return -1

# Time complexity: O(n)
# Space complexity: O(1) 
# Function definition
def LinearSearch(x, arr):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1

# Driver code
arr = [2, 1, 8, 9, 12, 15, 11, 19]
x = 20

# Function Definition
result = LinearSearch(x, arr)
print("Searching element is present at index", result)
```

```python
## Best time to buy and sell stock
# function definition
def findMaxProfit(prices):
    min_price = float('inf')
    max_profit = 0

    for i in range(len(prices)):
        if prices[i] < min_price:
            min_price = prices[i]
        elif prices[i] - min_price > max_profit:
            max_profit = prices[i] - min_price
    return max_profit



# Driver code
prices = [7, 5, 3, 6, 4, 15]
maxProfit = findMaxProfit(prices)
print("The maximum possible profit is", maxProfit)
```

```python
## Collinearity

# Aprroach 1: Using slope concept
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    if ((y2 - y1) * (x3 - x2) == (y3 - y2)*(x2 - x1)):
        return True
    return False


# Driver code
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)

```

```python
# Approach 2: Using area of triangle
def isCollinearPoints(x1, x2, x3, y1, y2, y3):
    area = 0.5 * (x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2))

    if area == 0:
        return True
    return False

# Driver code
x1, x2, x3, y1, y2, y3 = 1, 1, 1, 6, 0, 9
isCollinearPoints(x1, x2, x3, y1, y2, y3)
```

```python
## Majority Element - Boyer-Moore Voting Algorithm

# Function definition
def majorityElement(nums):
    # find candidate
    count = 0
    candidate = None
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)

    # validate majority status
    validation_count = 0
    for num in nums:
        if num == candidate:
            validation_count += 1
    
    if validation_count > len(nums)//2:
        return candidate
    return None
    


# Driver Code
# nums = [2, 2, 1, 1, 1, 2, 2]
nums = [2, 2, 7, 3, 4]
result = majorityElement(nums)
print("The majority element is", result)
```

```python
## Sort Colours -- Two-Pointer
# function definition
def sortColours(nums):
    curr = 0
    p1 = 0
    p2 = len(nums) - 1
    while curr < p2:
        if nums[curr] == 0:
            nums[curr], nums[p1] = nums[p1], nums[curr]
            p1 += 1
            curr += 1
        elif nums[curr] == 2:
            nums[curr], nums[p2] = nums[p2], nums[curr]
            p2 -= 1
            curr += 1
        else:
            curr += 1
    return nums

# Driver Code
nums = [2, 0, 2, 1, 1, 0]
result = sortColours(nums)
print(result)
```

```python
## Top K Frequent Elements
import heapq
def topKFrequent(nums, k):
    if len(nums) == k:
        return nums
    
    count = dict()
    min_heap = []
    for num in nums:
        if num in count:
            count[num] += 1
        else:
            count[num] = 1
    for key, value in count.items():
        heapq.heappush(min_heap, (value, key))
        if len(min_heap) > k:
            heapq.heappop(min_heap)

    return [num for _, num in min_heap]

# Driver Code
nums = [1,1,1,1,3,3,2,2,2,26,6,6,6,6,6,]
k = 2
print(topKFrequent(nums,k))
    
```

```python
## Longest Consecutive Sequence
def longestConsecutive(nums):
    if len(nums) == 0:
        return 0
    
    numSet = set(nums)
    longestSub = 1

    for num in numSet:
        if num-1 in numSet:
            continue
        else:
            currentNum = num
            currentSub = 1
            while currentNum+1 in numSet:
                currentNum+=1
                currentSub+=1
            longestSub = max(longestSub, currentSub)
    return longestSub

nums = [0, 3, 7, 2, 5, 8, 4, 6, 0, 1]
print(longestConsecutive(nums))

```

```python
# Contains Duplicate II
def containsNearbyDuplicate(nums: list[int], k: int) -> bool:
    hash_set = set()

    for idx in range(len(nums)):
        if nums[idx] in hash_set:
            return True
        else:
            hash_set.add(nums[idx])
        if len(hash_set) > k:
            hash_set.remove(nums[idx - k])

    return False

nums = [1,2,4,5,2]
k = 1
containsNearbyDuplicate(nums, k)
```

```python
# Product of Array Except Self
def productExceptSelf(nums: list):
    result = [1] * len(nums)
    pre, post = 1,1

    for i in range(len(nums)):
        result[i] = pre
        pre = nums[i] * pre

    for i in reversed(range(len(nums))):
        result[i] = result[i] * post
        post = post * nums[i]

    return result

productExceptSelf([1,2,3,4])
```

```python
# Top K Frequemt Elements
import heapq

def topKFrequent(nums: list[int], k: int):
    if k == len(nums):
        return nums

    count: dict[int, int] = dict()
    for num in nums:
        count[num] = count.get(num, 0) + 1

    heap = []
    for key, value in count.items():
        heapq.heappush(heap, (value, key))
        if len(heap) > k:
            heapq.heappop(heap)
    
    return [num for _, num in heap]

nums = [1,1,1,11,1,2,4,4,56,6,7,8,88,88,8,8,8,8,4,1,1,1,1,1,9,1,1,1]
k = 3
topKFrequent(nums, k)
```

```python
# Longest Consecutive Sequence
def longestConsecutive(nums: list[int]) -> int:
    if len(nums) == 1 or len(nums) == 0:
        return len(nums)
    nums_set = set(nums)
    lcs = 1
    
    for i in range(len(nums)):
        if nums[i] - 1 in nums_set:
            continue
        else:
            currNum = nums[i]
            cs = 1
            while currNum + 1 in nums_set:
                currNum+=1
                cs+=1
        if cs > lcs:
            lcs = cs
    return lcs

nums = [100, 4, 200, 1, 2, 3]
longestConsecutive(nums)
```

```python
# Best Time to Buy and Sell Stock
def maxProfit(prices: list[int]):
    min_p = prices[0]
    profit = 0

    for price in prices:
        if price < min_p:
            min_p = price

        if price - min_p > profit:
            profit = price - min_p

    return profit

prices = [7, 1, 5, 3, 6, 4]
maxProfit(prices)
```
