```python
# Implement Queue using Stacks
from typing import Generic, TypeVar

A = TypeVar('A')

class Stack(Generic[A]):
    def __init__(self) -> None:
        self.__items = list()

    def push(self, item: A) -> None:
        self.__items.append(item)

    def pop(self) -> A:
        item = self.__items[-1]
        self.__items.remove(item)
        return item

    def peek(self) -> A:
        if not self.isEmpty():
            return self.__items[-1]
        return None

    def isEmpty(self) -> bool:
        return True if len(self.__items) == 0 else False


class MyQueue:
    def __init__(self) -> None:
        self.__inStack = Stack()
        self.__outStack = Stack()

    def push(self, x: int) -> None:
        self.__inStack.push(x)

    def pop(self) -> int:
        self.moveIfNeeded()
        return self.__outStack.pop()

    def peek(self) -> int:
        self.moveIfNeeded()
        return self.__outStack.peek()

    def empty(self) -> bool:
        return self.__inStack.isEmpty() and self.__outStack.isEmpty()

    def moveIfNeeded(self) -> None:
        if self.__outStack.isEmpty():
            while not self.__inStack.isEmpty():
                self.__outStack.push(self.__inStack.pop())

queue = MyQueue()
queue.push(3)
queue.push(5)
print(queue.peek())
print(queue.pop())
queue.empty()
```

```python
# Implement Stack using Queues
from queue import Queue

class MyStack:
    def __init__(self) -> None:
        self.q1: Queue[int] = Queue()
        self.q2: Queue[int] = Queue()

    def push(self, x: int) -> None:
        self.q2.put(x)

        while not self.q1.empty():
            self.q2.put(self.q1.get())

        temp = self.q1
        self.q1 = self.q2
        self.q2 = temp

    def pop(self) -> int:
        return self.q1.get()

    def top(self) -> int:
        return self.q1.queue[0]

    def empty(self) -> bool:
        return self.q1.empty()


stack = MyStack()
stack.push(3)
stack.push(5)
print(stack.top())
print(stack.pop())
stack.empty()
```

```python
# Design Circular Queue
class MyCircularQueue:
    def __init__(self, k: int) -> None:
        self.data = [0] * k
        self.cap = k
        self.head = 0
        self.size = 0

    def enQueue(self, value: int) -> bool:
        if self.isFull():
            return False

        tail = (self.head + self.size) % self.cap
        self.data[tail] = value
        self.size += 1
        return True

    def deQueue(self) -> bool:
        if self.isEmpty():
            return False

        self.head = (self.head + 1) % self.cap
        self.size -= 1
        return True

    def Front(self) -> int:
        return -1 if self.isEmpty() else self.data[self.head]

    def Rear(self) -> int:
        if self.isEmpty():
            return -1
        tail = (self.head + self.size - 1) % self.cap
        return self.data[tail]

    def isEmpty(self) -> bool:
        return self.size == 0

    def isFull(self) -> bool:
        return self.size == self.cap
    
```

```python
# Open the Lock
from queue import Queue

def openLock(deadends: list[str], target: str) -> int:
    dead = set(deadends)
    if '0000' in dead:
        return -1
    if '0000' == target:
        return 0

    q: Queue[str] = Queue()
    seen = set()

    q.put('0000')
    seen.add('0000')

    steps = 0

    while not q.empty():
        size = len(q.queue)
        for s in range(size):
            cur: str = q.get()
            if cur in dead:
                continue
            if cur == target:
                return steps
            cs: list[str] = list(cur)

            for i in range(4):
                orig: str = cs[i]
                d: int = ord(orig) - ord('0')

                cs[i] = chr(ord('0') + ((d + 1) % 10))
                up: str = "".join(cs)
                if up not in dead and seen.add(up):
                    q.put(up)

                cs[i] = chr(ord('0') + ((d + 9) % 10))
                down: str = "".join(cs)
                if down not in dead and seen.add(down):
                    q.put(down)

    return -1
```

```python
# Dota2 Senate
from queue import Queue

def predictPartyVictory(senate: str):
    n = len(senate)
    r: Queue[int] = Queue()
    d: Queue[int] = Queue()

    for i in range(n):
        if senate[i] == 'R':
            r.put(i)
        else:
            d.put(i)

    while not r.empty() and not d.empty():
        ri = r.get()
        di = d.get()

        if ri < di:
            r.put(ri + n)
        else:
            d.put(di + n)

    return "Dire" if r.empty() else "Radiant"

senate = "RRRDDDDRRD"
predictPartyVictory(senate)
```

```python
# Heap Implementation
class MinHeap:
    def __init__(self, capacity: int) -> None:
        self.capacity = capacity
        self.size = 0
        self.heap = [0] * capacity

    def __parent(self, i: int) -> int:
        return (i-1)//2

    def __leftChild(self, i: int) -> int:
        return 2 * i + 1

    def __rightChild(self, i: int) -> int:
        return 2 * i + 2

    def __isLeaf(self, i: int) -> bool:
        return i >= self.size // 2 and i < self.size

    def insert(self, element: int) -> None:
        if self.size == self.capacity:
            raise Exception("Heap is full")

        self.heap[self.size] = element
        current = self.size
        self.size+=1

        while self.heap[current] < self.heap[self.__parent(current)]:
            self.swap(current, self.__parent(current))
            current = self.__parent(current)

    def removeMin(self) -> int:
        if self.size == 0:
            raise Exception("Heap is empty")

        mini = self.heap[0]
        self.size -= 1
        self.heap[0] = self.heap[self.size]
        self.heapify(0)

        return mini

    def heapify(self, i: int) -> None:
        if self.__isLeaf(i):
            return

        left = self.__leftChild(i)
        right = self.__rightChild(i)
        smallest = i

        if left < self.size and self.heap[left] < self.heap[i]:
            smallest = left

        if right < self.size and self.heap[right] < self.heap[smallest]:
            smallest = right

        if smallest != i:
            self.swap(i, smallest)
            self.heapify(smallest)

    def swap(self, i: int, j: int) -> None:
        temp = self.heap[i]
        self.heap[i] = self.heap[j]
        self.heap[j] = temp

min_heap = MinHeap(5)
min_heap.insert(5)
min_heap.insert(6)
min_heap.insert(12)
min_heap.insert(4)
min_heap.insert(25)
min_heap.removeMin()
```

```python
# Last Stone Weight
import heapq


def lastStoneWeight(stones: list[int]) -> int:
    maxHeap = [-stone for stone in stones]
    heapq.heapify(maxHeap)

    while len(maxHeap) > 1:
        y = -heapq.heappop(maxHeap)
        x = -heapq.heappop(maxHeap)

        if x != y:
            heapq.heappush(maxHeap, -(y-x))

    return -maxHeap[0] if maxHeap else 0

stones = [2, 7, 4, 1, 8, 1]
lastStoneWeight(stones)
```

```python
# Kth Largest Element in an Array
import heapq

def kthLargest(nums: list[int], k: int) -> int:
    nums = [-num for num in nums]
    heapq.heapify(nums)

    kth = int()
    for _ in range(k):
        kth = heapq.heappop(nums)

    return -kth

def alternative(nums: list[int], k: int) -> int:
    minHeap = []

    for num in nums:
        heapq.heappush(minHeap, num)
        if len(minHeap) > k:
            heapq.heappop(minHeap)

    return minHeap[0]

nums = [4, 2, 6, 1, 7, 3, 2]
k = 3

# kthLargest(nums, k)
alternative(nums, k)
```
