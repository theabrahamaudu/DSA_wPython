```python
# Implement Queue using Stacks
from typing import Generic, TypeVar

A = TypeVar('A')

class Stack(Generic[A]):
    def __init__(self) -> None:
        self.__items = list()

    def push(self, item: A) -> None:
        self.__items.append(item)

    def pop(self) -> A:
        item = self.__items[-1]
        self.__items.remove(item)
        return item

    def peek(self) -> A:
        if not self.isEmpty():
            return self.__items[-1]
        return None

    def isEmpty(self) -> bool:
        return True if len(self.__items) == 0 else False


class MyQueue:
    def __init__(self) -> None:
        self.__inStack = Stack()
        self.__outStack = Stack()

    def push(self, x: int) -> None:
        self.__inStack.push(x)

    def pop(self) -> int:
        self.moveIfNeeded()
        return self.__outStack.pop()

    def peek(self) -> int:
        self.moveIfNeeded()
        return self.__outStack.peek()

    def empty(self) -> bool:
        return self.__inStack.isEmpty() and self.__outStack.isEmpty()

    def moveIfNeeded(self) -> None:
        if self.__outStack.isEmpty():
            while not self.__inStack.isEmpty():
                self.__outStack.push(self.__inStack.pop())

queue = MyQueue()
queue.push(3)
queue.push(5)
print(queue.peek())
print(queue.pop())
queue.empty()
```

```python
# Implement Stack using Queues
from queue import Queue

class MyStack:
    def __init__(self) -> None:
        self.q1: Queue[int] = Queue()
        self.q2: Queue[int] = Queue()

    def push(self, x: int) -> None:
        self.q2.put(x)

        while not self.q1.empty():
            self.q2.put(self.q1.get())

        temp = self.q1
        self.q1 = self.q2
        self.q2 = temp

    def pop(self) -> int:
        return self.q1.get()

    def top(self) -> int:
        return self.q1.queue[0]

    def empty(self) -> bool:
        return self.q1.empty()


stack = MyStack()
stack.push(3)
stack.push(5)
print(stack.top())
print(stack.pop())
stack.empty()
```

```python
# Design Circular Queue
class MyCircularQueue:
    def __init__(self, k: int) -> None:
        self.data = [0] * k
        self.cap = k
        self.head = 0
        self.size = 0

    def enQueue(self, value: int) -> bool:
        if self.isFull():
            return False

        tail = (self.head + self.size) % self.cap
        self.data[tail] = value
        self.size += 1
        return True

    def deQueue(self) -> bool:
        if self.isEmpty():
            return False

        self.head = (self.head + 1) % self.cap
        self.size -= 1
        return True

    def Front(self) -> int:
        return -1 if self.isEmpty() else self.data[self.head]

    def Rear(self) -> int:
        if self.isEmpty():
            return -1
        tail = (self.head + self.size - 1) % self.cap
        return self.data[tail]

    def isEmpty(self) -> bool:
        return self.size == 0

    def isFull(self) -> bool:
        return self.size == self.cap
    
```
