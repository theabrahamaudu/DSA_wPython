```python
## Best time to buy and sell stock
def maxProfit(prices):
    min_price = prices[0]
    profit = 0

    for price in prices:
        if price < min_price:
            min_price = price

        profit = max(profit, price - min_price)

    return profit


prices = [7, 1, 5, 3, 6, 4]
print(maxProfit(prices))
```

```python
## Permutation in String
from array import array

def matches(s1Map, s2Map):
    for i in range(26):
        if s1Map[i] != s2Map[i]:
            return False
        
    return True

def checkInclusion(s1: str, s2: str):
    if len(s1) > len(s2):
        return False
    
    s1Map = array('i', [0]*26)
    s2Map = array('i', [0]*26)

    for i in range(len(s1)):
        s1Map[ord(s1[i]) - ord("a")]+=1
        s2Map[ord(s2[i]) - ord("a")]+=1

    for i in range(len(s2) - len(s1)):
        if matches(s1Map, s2Map):
            return True
        s2Map[ord(s2[i + len(s1)]) - ord("a")]+=1
        s2Map[ord(s2[i]) - ord("a")]-=1
    
    return matches(s1Map, s2Map)

s1 = "aabapl"
s2 = "malaapalonio"
print(checkInclusion(s1, s2))
```

```python
## Longest repeating character replacement
from array import array
def characterReplacement(s, k):
    occurence = array('i', [0]*26)
    left, right = 0, 0
    ans = 0
    maxOccurence = 0

    for right in range(len(s)):
        occurence[ord(s[right]) - ord('A')]+=1
        maxOccurence = max(maxOccurence, occurence[ord(s[right]) - ord('A')])
        if right - left + 1 - maxOccurence > k:
            occurence[ord(s[left]) - ord('A')] -= 1
            left+=1
        ans = max(ans, right - left + 1)

    return ans


s = "ABABA"
k = 2

print(characterReplacement(s,k))

```

```python
## Longest Substring without repeating characters
def lengthOfLongestSubstring(s: str):
    if s == None or len(s) == 0:
        return 0
    
    if len(s) == 1:
        return 1
    
    left = 0
    right = 0
    ans = 0

    Set = set()

    while right < len(s):
        c = ord(s[right])
        while c in Set:
            Set.remove(ord(s[left]))
            left+=1
        Set.add(c)
        ans = max(ans, right - left + 1)
        right+=1

    return ans


s = "abcabcbb"
print(lengthOfLongestSubstring(s))


```
