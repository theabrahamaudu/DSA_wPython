```python
## Bubble Sort
# Funtion implementation
def bubbleSort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr




# Driver code
arr = [70, 20, 50, 30, 90, 5, 15]
# Function call
result = bubbleSort(arr)
print(result)
```

```python
## Selection sort
# function definition
def selectionSort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


# Driver code
arr = [50, 38, 45, 79, 19, 27, 29]
result = selectionSort(arr)
print(result)
```

```python
## Insertion Sort
# Function Definition
def insertionSort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]
            j = j-1
        arr[j+1] = key
    return arr



# Dtiver Code
arr = [9, 5, 1, 4, 3]

# Function call
result = insertionSort(arr)
print("The sorted array is", result)
```
