```python
## Bubble Sort
# Funtion implementation
def bubbleSort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr




# Driver code
arr = [70, 20, 50, 30, 90, 5, 15]
# Function call
result = bubbleSort(arr)
print(result)
```

```python
## Selection sort
# function definition
def selectionSort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


# Driver code
arr = [50, 38, 45, 79, 19, 27, 29]
result = selectionSort(arr)
print(result)
```

```python
## Insertion Sort
# Function Definition
def insertionSort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]
            j = j-1
        arr[j+1] = key
    return arr



# Dtiver Code
arr = [9, 5, 1, 4, 3]

# Function call
result = insertionSort(arr)
print("The sorted array is", result)
```

```python
# Binary Search
def binarySearch(nums: list[int], target: int):
    left = 0
    right =  len(nums) - 1

    while left <= right:
        mid = left + (right - left)//2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            right = mid-1
        else:
            left = mid+1
    return -1

nums = [1, 2, 3, 4, 5, 6, 7]
binarySearch(nums, 5)
```

```python
# First and Last Positon of Elements in a Sorted Array
def searchRange(nums: list[int], target: int) -> list[int]:
    first = findBound(nums, target, True)

    if first == -1:
        return [-1, -1]

    last = findBound(nums, target, False)
    return [first, last]


def findBound(nums: list[int], target: int, isFirst: bool) -> int:
    start = 0
    end = len(nums) - 1

    while start <= end:
        mid  = start + (end - start)//2
        if nums[mid] == target:
            if isFirst:
                if mid == start or nums[mid-1] != target:
                    return mid

                end = mid - 1
            else:
                if mid == end or nums[mid+1] != target:
                    return mid
                start = mid + 1

        elif nums[mid] > target:
            end = mid - 1
        else:
            start = mid + 1
    return -1

nums = [1,2,3,3,3,3,4,5,]
searchRange(nums, 3)
```

```python
# Merge Sorted Array
def merge(nums1: list[int], m: int, nums2: list[int], n):
    p1 = m-1; p2 = n-1; i = m + n - 1

    while p2 >= 0:
        if p1 >= 0 and nums1[p1] > nums2[p2]:
            nums1[i] = nums1[p1]
            p1 -= 1
        else:
            nums1[i] = nums2[p2]
            p2 -= 1
        i -= 1

    return nums1

nums1 = [1,3,5,0,0,0]
nums2 = [2,4,6]
merge(nums1, 3, nums2, 3)
```

```python
# Sort Colours
def sortColours(nums: list[int]):
    low = 0
    high = len(nums) - 1
    curr = 0

    while curr <= high:
        if nums[curr] == 0:
            swap(nums, curr, low)
            low +=1
            curr+=1

        elif nums[curr] == 2:
            swap(nums, curr, high)
            high -=1
        else:
            curr+=1
    return nums

def swap(nums: list[int], i: int, j: int):
    temp = nums[i]
    nums[i] = nums[j]
    nums[j] = temp

nums = [1,1,1,2,2,0,0,2]
sortColours(nums)
```
