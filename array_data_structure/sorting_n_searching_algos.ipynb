```python
## Bubble Sort
# Funtion implementation
def bubbleSort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr




# Driver code
arr = [70, 20, 50, 30, 90, 5, 15]
# Function call
result = bubbleSort(arr)
print(result)
```

```python
## Selection sort
# function definition
def selectionSort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


# Driver code
arr = [50, 38, 45, 79, 19, 27, 29]
result = selectionSort(arr)
print(result)
```

```python
## Insertion Sort
# Function Definition
def insertionSort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]
            j = j-1
        arr[j+1] = key
    return arr



# Dtiver Code
arr = [9, 5, 1, 4, 3]

# Function call
result = insertionSort(arr)
print("The sorted array is", result)
```

```python
# Binary Search
def binarySearch(nums: list[int], target: int):
    left = 0
    right =  len(nums) - 1

    while left <= right:
        mid = left + (right - left)//2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            right = mid-1
        else:
            left = mid+1
    return -1

nums = [1, 2, 3, 4, 5, 6, 7]
binarySearch(nums, 5)
```

```python
# First and Last Positon of Elements in a Sorted Array
def searchRange(nums: list[int], target: int) -> list[int]:
    first = findBound(nums, target, True)

    if first == -1:
        return [-1, -1]

    last = findBound(nums, target, False)
    return [first, last]


def findBound(nums: list[int], target: int, isFirst: bool) -> int:
    start = 0
    end = len(nums) - 1

    while start <= end:
        mid  = start + (end - start)//2
        if nums[mid] == target:
            if isFirst:
                if mid == start or nums[mid-1] != target:
                    return mid

                end = mid - 1
            else:
                if mid == end or nums[mid+1] != target:
                    return mid
                start = mid + 1

        elif nums[mid] > target:
            end = mid - 1
        else:
            start = mid + 1
    return -1

nums = [1,2,3,3,3,3,4,5,]
searchRange(nums, 3)
```

```python
# Merge Sorted Array
def merge(nums1: list[int], m: int, nums2: list[int], n):
    p1 = m-1; p2 = n-1; i = m + n - 1

    while p2 >= 0:
        if p1 >= 0 and nums1[p1] > nums2[p2]:
            nums1[i] = nums1[p1]
            p1 -= 1
        else:
            nums1[i] = nums2[p2]
            p2 -= 1
        i -= 1

    return nums1

nums1 = [1,3,5,0,0,0]
nums2 = [2,4,6]
merge(nums1, 3, nums2, 3)
```

```python
# Sort Colours
def sortColours(nums: list[int]):
    low = 0
    high = len(nums) - 1
    curr = 0

    while curr <= high:
        if nums[curr] == 0:
            swap(nums, curr, low)
            low +=1
            curr+=1

        elif nums[curr] == 2:
            swap(nums, curr, high)
            high -=1
        else:
            curr+=1
    return nums

def swap(nums: list[int], i: int, j: int):
    temp = nums[i]
    nums[i] = nums[j]
    nums[j] = temp

nums = [1,1,1,2,2,0,0,2]
sortColours(nums)
```

```python
# Majority Element
def countNums(nums: list[int]) -> dict[int, int]:
    counts: dict[int, int] = dict()

    for num in nums:
        if num not in counts.keys():
            counts[num] = 1
        else:
            counts[num] = counts[num]+1
    return counts

def majorityElement(nums: list[int]) -> int:
    counts = countNums(nums)

    majorityEntry: dict[int, int] = dict()
    for key, value in counts.items():
        if not majorityEntry or value > next(iter(majorityEntry.values())):
            majorityEntry = {key: value}
    return next(iter(majorityEntry))

nums = [2,1,1,1,2,2,2]
majorityElement(nums)
```

```python
# Search in a 2D Matrix
def searchMatrix(matrix: list[list[int]], target: int):
    m = len(matrix)
    n = len(matrix[0])

    left = 0
    right = m*n - 1

    while left <= right:
        mid = left + (right - left)//2
        midValue = matrix[mid // n][mid % n]

        if midValue == target:
            return True
        elif midValue > target:
            right = mid - 1
        else:
            left = mid + 1
    return False

matrix = [
    [1, 3, 7],
    [10, 14, 17],
    [20, 34, 60]
]
target = 30
searchMatrix(matrix, target)
```

```python
# Find Minimum in a Rotated Sorted Array
def findMin(nums: list[int]):
    left = 0
    right = len(nums)-1
    ans = nums[0]

    if len(nums) == 1:
        return ans

    while left <= right:
        
        if nums[left] < nums[right]:
            ans = min(ans, nums[left])

        mid = left + (right-left)//2
        ans = min(ans, nums[mid])

        if nums[left] <= nums[mid]:
            left = mid + 1

        else:
            right = mid - 1

    return ans

nums = [2, 3, 1]
findMin(nums)
```

```python
# Search in a Rotated Sorted Array
def search(nums: list[int], target: int):
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = left + (right-left)//2

        if nums[mid] == target:
            return mid

        if nums[left] <= nums[mid]:
            if target < nums[left] or target > nums[mid]:
                left = mid + 1
            else:
                right = mid - 1
        else:
            if target > nums[right] or target < nums[mid]:
                right = mid - 1
            else:
                left = mid + 1
    return -1

nums = [9, 1, 2, 4, 5, 7, 8]
search(nums, 2)
```

```python
# Sort List
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next: ListNode = None

def sortList(head: ListNode):
    if head == None or head.next == None:
        return head

    mid: ListNode = getMid(head)
    left: ListNode = sortList(head)
    right: ListNode = sortList(mid)

    return merge(left, right)

def getMid(head: ListNode) -> ListNode:
    prev: ListNode = None

    while head != None and head.next != None:
        prev = head if prev is None else prev.next
        head = head.next.next

    mid: ListNode = prev.next
    prev.next = None
    return mid

def merge(list1: ListNode, list2: ListNode) -> ListNode:
    dummy: ListNode = ListNode(0)
    tail: ListNode = dummy

    while list1 != None and list2 != None:
        if list1.data < list2.data:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next

        tail = tail.next

    tail.next = list1 if list1 != None else list2
    return dummy.next

def getList(head: ListNode) -> list:
    vals = []
    while head:
        vals.append(head.data)
        head = head.next
    return vals


node1 = ListNode(4)
node2 = ListNode(2)
node3 = ListNode(1)
node4 = ListNode(3)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1

sortedList = sortList(head)
getList(sortedList)
```

```python
# Largest Number
from functools import cmp_to_key

def largestNumber(nums: list[int]):
    asStrs: list[str] = []

    for i in range(len(nums)):
        asStrs.append(str(nums[i]))

    asStrs.sort(key=cmp_to_key(compare))

    if asStrs[0] == "0":
        return "0"

    largestNumberStr = str()
    for numAsStr in asStrs:
        largestNumberStr+=numAsStr

    return largestNumberStr
    


def compare(a: str, b: str) -> int:
    order1 = a + b
    order2 = b + a

    if order2 < order1:
        return -1
    elif order2 > order1:
        return 1
    else:
        return 0

nums = [10, 2] 
largestNumber(nums)
```

```python
# Koko Eating Bananas
from math import ceil


def minEatingSpeed(piles: list[int], h: int) -> int:
    left, right = 1, 1

    for pile in piles:
        right = max(pile, right)

    while left < right:
        mid = left + (right - left)//2
        if canFinish(piles, mid, h):
            right = mid
        else:
            left = mid+1
    return left

def canFinish(piles: list[int], speed: int, h: int) -> bool:
    hours = 0
    for pile in piles:
        hours += ceil(pile/speed)

    return hours <= h

piles = [3, 6, 7, 11]
h = 8
minEatingSpeed(piles, h)
```

```python
# Time-Based Key-Value Store
from bisect import bisect_right
from collections import defaultdict

class TimeMap:
    def __init__(self):
        self.map: dict[str, tuple[list[int], list[str]]] = defaultdict(lambda: ([], []))

    def set(self, key: str, value: str, timestamp: int) -> None:
        timestamps, values = self.map[key]

        timestamps.append(timestamp)
        values.append(value)

    def get(self, key: str, timestamp: int) -> str:
        if key not in self.map:
            return ""

        timestamps, values = self.map[key]
        i = bisect_right(timestamps, timestamp)
        if i == 0:
            return ""
        return values[i-1]


vals = [("a", "abraham", 1), ("b", "Bonita", 2), ("b", "Bonita", 3)]
obj = TimeMap()

for val in vals:
    obj.set(val[0], val[1], val[2])

obj.get("b", 3)
```
