```python
# Stack Implementation
from typing import Generic, TypeVar

A = TypeVar('A')
class Stack(Generic[A]):
    def __init__(self):
        self.__items = list()

    def push(self, item: A) -> None:
        self.__items.append(item)

    def pop(self) -> int:
        if not self.isEmpty():
            item = self.__items[-1]
            self.__items.remove(item)
            return item
        return None

    def peek(self) -> int:
        if not self.isEmpty():
            item = self.__items[-1]
            return item
        return None

    def isEmpty(self) -> bool:
        return True if len(self.__items) == 0 else False      
```

```python
stack = Stack()

stack.push(10)
stack.push(20)
stack.push(30)
```

```python
print("top element:", stack.peek())
```

```python
print("popped element:", stack.pop())
print("popped element:", stack.pop())
```

```python
print("top element after popping:", stack.peek())
```

```python
# Valid Parenthesis
def isValid(s: str) -> bool:
    mappedBrackets: dict[str, str] = dict()

    mappedBrackets[')'] = '('
    mappedBrackets['}'] = '{'
    mappedBrackets[']'] = '['

    stack: Stack[str] = Stack()

    for char in s:
        if char not in mappedBrackets.keys():
            stack.push(char)
        else:
            if stack.isEmpty():
                return False
            topElement = stack.pop()
            if topElement != mappedBrackets.get(char):
                return False
    return True if stack.isEmpty() else False

s = "([])"
isValid(s)
```

```python
# Min Stack
from typing import Any

class Node:
    def __init__(self, val: int, mini: int, next: Any) -> None:
        self.val = val
        self.mini: int = mini
        self.next: Node = next

class MinStack:
    def __init__(self) -> None:
        self.__head: Node = None

    def push(self, val: int) -> None:
        if self.__head == None:
            self.__head = Node(val, val, None)
        else:
            self.__head = Node(
                val,
                min(val, self.__head.mini),
                self.__head
            )

    def pop(self) -> int:
        val = self.__head.val
        self.__head = self.__head.next
        return val

    def top(self) -> int:
        return self.__head.val

    def getMin(self):
        return self.__head.mini


stack = MinStack()
stack.push(10)
stack.push(2)
stack.push(9)
stack.push(3)
stack.push(13)
print(stack.top())
print(stack.pop())
print(stack.pop())
print(stack.getMin())


```

```python
# Max Stack
class MaxStack:
    def __init__(self) -> None:
        self.stack: Stack[int] = Stack()
        self.maxStack: Stack[int] = Stack()

    def push(self, x: int) -> None:
        maxi = x if self.maxStack.isEmpty() else max(self.maxStack.peek(), x)
        self.stack.push(x)
        self.maxStack.push(maxi)

    def pop(self) -> int:
        self.maxStack.pop()
        return self.stack.pop()

    def top(self) -> int:
        return self.stack.peek()

    def peekMax(self) -> int:
        return self.maxStack.peek()

    def popMax(self) -> int:
        maxi = self.peekMax()
        buffer: Stack[int] = Stack()
        while self.top() != maxi:
            buffer.push(self.pop())
        self.pop()
        while not buffer.isEmpty():
            self.push(buffer.pop())

        return maxi
```
