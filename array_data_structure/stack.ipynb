```python
# Stack Implementation
from typing import Generic, TypeVar

A = TypeVar('A')
class Stack(Generic[A]):
    def __init__(self):
        self.__items = list()

    def push(self, item: A) -> None:
        self.__items.append(item)

    def pop(self) -> int:
        if not self.isEmpty():
            item = self.__items[-1]
            self.__items.remove(item)
            return item
        return None

    def peek(self) -> int:
        if not self.isEmpty():
            item = self.__items[-1]
            return item
        return None

    def isEmpty(self) -> bool:
        return True if len(self.__items) == 0 else False      
```

```python
# Valid Parenthesis
def isValid(s: str) -> bool:
    mappedBrackets: dict[str, str] = dict()

    mappedBrackets[')'] = '('
    mappedBrackets['}'] = '{'
    mappedBrackets[']'] = '['

    stack: Stack[str] = Stack()

    for char in s:
        if char not in mappedBrackets.keys():
            stack.push(char)
        else:
            if stack.isEmpty():
                return False
            topElement = stack.pop()
            if topElement != mappedBrackets.get(char):
                return False
    return True if stack.isEmpty() else False

s = "([])"
isValid(s)
```

```python
# Min Stack
from typing import Any

class Node:
    def __init__(self, val: int, mini: int, next: Any) -> None:
        self.val = val
        self.mini: int = mini
        self.next: Node = next

class MinStack:
    def __init__(self) -> None:
        self.__head: Node = None

    def push(self, val: int) -> None:
        if self.__head == None:
            self.__head = Node(val, val, None)
        else:
            self.__head = Node(
                val,
                min(val, self.__head.mini),
                self.__head
            )

    def pop(self) -> int:
        val = self.__head.val
        self.__head = self.__head.next
        return val

    def top(self) -> int:
        return self.__head.val

    def getMin(self):
        return self.__head.mini


stack = MinStack()
stack.push(10)
stack.push(2)
stack.push(9)
stack.push(3)
stack.push(13)
print(stack.top())
print(stack.pop())
print(stack.pop())
print(stack.getMin())


```

```python
# Max Stack
class MaxStack:
    def __init__(self) -> None:
        self.stack: Stack[int] = Stack()
        self.maxStack: Stack[int] = Stack()

    def push(self, x: int) -> None:
        maxi = x if self.maxStack.isEmpty() else max(self.maxStack.peek(), x)
        self.stack.push(x)
        self.maxStack.push(maxi)

    def pop(self) -> int:
        self.maxStack.pop()
        return self.stack.pop()

    def top(self) -> int:
        return self.stack.peek()

    def peekMax(self) -> int:
        return self.maxStack.peek()

    def popMax(self) -> int:
        maxi = self.peekMax()
        buffer: Stack[int] = Stack()
        while self.top() != maxi:
            buffer.push(self.pop())
        self.pop()
        while not buffer.isEmpty():
            self.push(buffer.pop())

        return maxi
```

```python
# Daily Temperatures
def dailyTemperatures(temperatures: list[int]):
    n = len(temperatures)
    answer = [0] * n
    stack = Stack()

    for i in range(n):
        while not stack.isEmpty() and temperatures[i] > temperatures[stack.peek()]:
            index = stack.pop()
            answer[index] = i - index

        stack.push(i)
    return answer

temperatures = [79, 69, 68, 62, 74, 76, 80]
dailyTemperatures(temperatures)
```

```python
# Car Fleet
def carFleet(target: int, position: list[int], speed: list[int]) -> int:
    n = len(position)
    cars: list[list[int]] = [[0, 0] for _ in range(n)]

    for i in range(n):
        cars[i][0] = position[i]
        cars[i][1] = (target - position[i])//speed[i]

    cars.sort(key=lambda x: x[0], reverse=True)

    count = 0
    prevTime = 0

    for car in cars:
        if car[1] > prevTime:
            count += 1
            prevTime = car[1]

    return count

target = 12
position = [10, 8, 0, 5, 3]
speed = [5, 1, 1, 10, 1]
carFleet(target, position, speed)
```

```python
# Evaluate Reverse Polish Notation
def evalRPN(tokens: list[str]):
    stack = Stack()

    for token in tokens:
        if isOperator(token):
            b = stack.pop()
            a = stack.pop()
            result = applyOperator(token, a, b)
            stack.push(result)
        else:
            stack.push(int(token))

    return stack.pop()

def isOperator(token: str) -> bool:
    return token == '+' or token == '-' or token == '*' or token == '/'

def applyOperator(operator: str, a: int, b: int):
    match operator:
        case '+':
            return a + b
        case '-':
            return a - b
        case '*':
            return a * b
        case '/':
            return a / b
        case _:
            raise Exception("invalid operator")
    
    
tokens = ['1', '2', '+', '3', '*']
evalRPN(tokens)
```

```python
# Minimum Remove to Make Valid Parenthesis
def minRemoveToMakeValid(s: str):
    removeIndices: set[int] = set()
    stack: Stack[str] = Stack()

    for i in range(len(s)):
        c = s[i]

        if c == '(':
            stack.push(i)
        elif c == ')':
            if stack.isEmpty():
                removeIndices.add(i)
            else:
                stack.pop()

    while not stack.isEmpty():
        removeIndices.add(stack.pop())

    sb = str()
    for i in range(len(s)):
        if i not in removeIndices:
            sb += s[i]
    return sb

s = "ab)(gh))"
minRemoveToMakeValid(s)


```

```python
# Generate Parentheses
def generateParenthesis(n: int):
    ans = list()
    backtrack(ans, str(), 0, 0, n)
    return ans


def backtrack(ans: list[str], cur: str, open: int, close: int, max: int) -> None:
    if len(cur) == max*2:
        ans.append(cur)
        return

    if open < max:
        cur += '('
        backtrack(ans, cur, open+1, close, max)
        cur = cur[:-1]

    if close < open:
        cur += ')'
        backtrack(ans, cur, open, close+1, max)
        cur  = cur[:-1]

n=4
generateParenthesis(n)
```
