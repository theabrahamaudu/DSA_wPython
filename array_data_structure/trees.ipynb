```python
# Maximum Depth of Binary Tree
class TreeNode:
    def __init__(self, val: int) -> None:
        self.val = val
        self.left: TreeNode = None
        self.right: TreeNode = None


def maxDepth(root: TreeNode) -> int:
    if root == None:
        return 0
    else:
        leftmax = maxDepth(root.left)
        rightmax = maxDepth(root.right)
        return max(leftmax, rightmax) + 1

node1 = TreeNode(1)
node2 = TreeNode(3)
node3 = TreeNode(2)
node4 = TreeNode(4)

node1.left = node2
node2.left = node3
node1.right = node4

maxDepth(node1)
```

```python
# Same Tree
def isSameTree(p: TreeNode, q: TreeNode) -> bool:
    if p is None and q is None:
        return True
    if p is None or q is None:
        return False
    if p.val != q.val: 
        return False

    return isSameTree(p.left, q.left) and isSameTree(p.right, q.right)


node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(1)
node4 = TreeNode(2)

node1.left = node2
node3.left = node4

isSameTree(node1, node3)
```

```python
# Symmetric Tree
class Solution:
    def isSymmetric(self, root: TreeNode) -> bool:
        if root.val is None:
            return True
        return self.__isMirror(root.left, root.right)

    def __isMirror(self, left: TreeNode, right: TreeNode) -> bool:
        if left is None and right is None:
            return True
        if left is None or left is None:
            return False
        return left.val == right.val\
            and self.__isMirror(left.left, right.right)\
            and self.__isMirror(left.right, right.left)

node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(2)
node4 = TreeNode(3)
node5 = TreeNode(3)

node1.left = node2
node1.right = node3
node2.left = node4
node3.right = node5

Solution().isSymmetric(node1)

```

```python
# Invert Binary Tree
def invertTree(root: TreeNode) -> TreeNode:
    if root is None:
        return None

    left = invertTree(root.right)
    right = invertTree(root.left)

    root.right = left
    root.left = right
    return root


node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(2)
node4 = TreeNode(3)
node5 = TreeNode(3)

node1.left = node2
node1.right = node3
node2.left = node4
node3.right = node5

invertTree(node1)
```

```python
# Balanced Binary Tree


class Solution:
    def isBalanced(self, root: TreeNode) -> bool:
        if root is None:
            return True

        leftHeight: int = self.__getHeight(root.right)
        rightHeight: int = self.__getHeight(root.left)

        if abs(leftHeight -rightHeight) > 1:
            return False

        return self.isBalanced(root.left) and self.isBalanced(root.right)

    def __getHeight(self, node: TreeNode) -> int:
        if node is None:
            return 0

        leftHeight: int = self.__getHeight(node.left)
        rightHeight: int = self.__getHeight(node.right)

        return max(leftHeight, rightHeight) + 1

node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(2)
node4 = TreeNode(3)
node5 = TreeNode(3)

node1.left = node2
node1.right = node3
node2.left = node4
node3.right = node5

Solution().isBalanced(node1)
```

```python
# Binary Tree Level Order Traversal
class Solution:
    def __init__(self) -> None:
        self.ans: list[list[int]] = list()

    def order(self, node: TreeNode, level: int) -> None:
        if len(self.ans) == level:
            self.ans.append(list())
        self.ans[level].append(node.val)

        if node.left is not None:
            self.order(node.left, level+1)

        if node.right is not None:
            self.order(node.right, level+1)

    def levelOrder(self, root: TreeNode) -> list[list[int]]:
        if root is None:
            return self.ans

        self.order(root, 0)
        return self.ans

tree = [3,9,20,None,None,15,7]

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

Solution().levelOrder(node1)
```

```python
# Binary Tree ZigZag Level Order Traversal
from collections import deque

def zigzagLevelOrder(root: TreeNode) -> list[list[int]]:
    if root is None:
        return []

    results: list[list[int]] = list()
    node_queue = deque()
    node_queue.append(root)
    node_queue.append(None)

    level_list = deque()
    is_order_left = True

    while len(node_queue) > 0:
        curr_node: TreeNode = node_queue.popleft()
        if curr_node is not None:
            if is_order_left:
                level_list.append(curr_node.val)
            else:
                level_list.appendleft(curr_node.val)

            if curr_node.left is not None:
                node_queue.append(curr_node.left)
            if curr_node.right is not None:
                node_queue.append(curr_node.right)

        else:
            results.append(list(level_list))
            level_list = deque()

            if len(node_queue) > 0:
                node_queue.append(None)
            is_order_left = not is_order_left

    return results


    
node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

zigzagLevelOrder(node1)



```

```python
# Binary Tree Right Side View
from collections import deque

def rightSideView(root: TreeNode) -> list[int]:
    result = list()

    if root is None:
        return result

    queue: deque[TreeNode] = deque()
    queue.append(root)

    while len(queue) > 0:
        levelSize = len(queue)

        for i in range(levelSize):
            currentNode = queue.popleft()

            if i == levelSize - 1:
                result.append(currentNode.val)

            if currentNode.left is not None:
                queue.append(currentNode.left)

            if currentNode.right is not None:
                queue.append(currentNode.right)

    return result

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

rightSideView(node1)
```

```python
# Subtree of Another Tree
def isSame(s: TreeNode, t: TreeNode) -> bool:
    if s is None and t is None:
        return True

    if s is None or t is None:
        return False

    if s.val != t.val:
        return False

    return isSame(s.left, t.left) and isSame(s.right, t.right)

def isSubtree(root: TreeNode, subRoot: TreeNode) -> bool:
    if root is None:
        return False

    if isSame(root, subRoot):
        return True

    return isSubtree(root.left, subRoot) or isSubtree(root.right, subRoot)

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

isSubtree(node1, node3)
```

```python
# Count Good Nodes in Binary Tree
def countGoodNodes(node: TreeNode, maxSoFar: int) -> int:
    if node is None:
        return 0

    count = 0
    if node.val >= maxSoFar:
        count += 1
        maxSoFar = node.val

    count += countGoodNodes(node.left, maxSoFar)
    count += countGoodNodes(node.right, maxSoFar)

    return count

def goodNodes(root: TreeNode) -> int:
    return countGoodNodes(root, float('-inf'))

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

goodNodes(node1)
```

```python
class Solution:
    def __init__(self) -> None:
        self.maxDiameter = 0

    def diameterOfBinaryTree(self, root: TreeNode) -> int:
        self.__getHeight(root)
        return self.maxDiameter

    def __getHeight(self, node: TreeNode) -> int:
        if node is None:
            return 0

        leftHeight = self.__getHeight(node.left)
        rightHeight = self.__getHeight(node.right)

        self.maxDiameter = max(self.maxDiameter, leftHeight + rightHeight)

        return 1 + max(leftHeight, rightHeight)

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

Solution().diameterOfBinaryTree(node1)
```

```python
# Validate Binary Search Tree
class Solution:
    def __init__(self) -> None:
        self.prev: int = None

    def isValidBST(self, root: TreeNode) -> bool:
        return self.inOrder(root)
    
    def inOrder(self,root: TreeNode) -> bool:
        if root is None:
            return True

        if not self.inOrder(root.left):
            return False

        if self.prev is not None and root.val <= self.prev:
            return False

        self.prev = root.val

        return self.inOrder(root.right)

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

Solution().isValidBST(node1)
```

```python
# Lowest Common Ancestor of a Binary Search Tree
def lowestCommonAncestor(root: TreeNode, p: TreeNode, q: TreeNode) -> TreeNode:
    parentVal = root.val
    pVal = p.val
    qVal = q.val

    if pVal > parentVal and qVal > parentVal:
        return lowestCommonAncestor(root.right, p, q)

    if pVal < parentVal and qVal < parentVal:
        return lowestCommonAncestor(root.left, p, q)

    else:
        return root

node1 = TreeNode(10)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(27)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

lowestCommonAncestor(node1, node6, node7).val
```

```python
# Kth Smallest Element in a BST
def inOrder(root: TreeNode, arr: list[int]) -> list[int]:
    if root is None:
        return arr

    inOrder(root.left, arr)
    arr.append(root.val)
    inOrder(root.right, arr)
    return arr

def kthSmallest(root: TreeNode, k: int) -> int:
    nums = inOrder(root, list[int]())
    return nums[k-1]

node1 = TreeNode(10)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(27)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

kthSmallest(node1, 1)

```

```python
# Find Leaves of Binary Tree
def removeLeaves(node: TreeNode, leaves: list[int]) -> TreeNode:
    if node is None:
        return None

    if node.left is None and node.right is None:
        leaves.append(node.val)
        return None

    node.left = removeLeaves(node.left, leaves)
    node.right = removeLeaves(node.right, leaves)

    return node

def collectLeaves(root: TreeNode) -> list[list[int]]:
    result: list[list[int]] = list()

    while root is not None:
        leaves: list[int] = list()
        root = removeLeaves(root, leaves)
        result.append(leaves)

    return result

node1 = TreeNode(10)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(27)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

collectLeaves(node1)
```
