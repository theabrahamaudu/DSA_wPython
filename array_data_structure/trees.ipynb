```python
# Maximum Depth of Binary Tree
class TreeNode:
    def __init__(self, val: int) -> None:
        self.val = val
        self.left: TreeNode = None
        self.right: TreeNode = None


def maxDepth(root: TreeNode) -> int:
    if root == None:
        return 0
    else:
        leftmax = maxDepth(root.left)
        rightmax = maxDepth(root.right)
        return max(leftmax, rightmax) + 1

node1 = TreeNode(1)
node2 = TreeNode(3)
node3 = TreeNode(2)
node4 = TreeNode(4)

node1.left = node2
node2.left = node3
node1.right = node4

maxDepth(node1)
```

```python
# Same Tree
def isSameTree(p: TreeNode, q: TreeNode) -> bool:
    if p is None and q is None:
        return True
    if p is None or q is None:
        return False
    if p.val != q.val: 
        return False

    return isSameTree(p.left, q.left) and isSameTree(p.right, q.right)


node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(1)
node4 = TreeNode(2)

node1.left = node2
node3.left = node4

isSameTree(node1, node3)
```

```python
# Symmetric Tree
class Solution:
    def isSymmetric(self, root: TreeNode) -> bool:
        if root.val is None:
            return True
        return self.__isMirror(root.left, root.right)

    def __isMirror(self, left: TreeNode, right: TreeNode) -> bool:
        if left is None and right is None:
            return True
        if left is None or left is None:
            return False
        return left.val == right.val\
            and self.__isMirror(left.left, right.right)\
            and self.__isMirror(left.right, right.left)

node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(2)
node4 = TreeNode(3)
node5 = TreeNode(3)

node1.left = node2
node1.right = node3
node2.left = node4
node3.right = node5

Solution().isSymmetric(node1)

```

```python
# Invert Binary Tree
def invertTree(root: TreeNode) -> TreeNode:
    if root is None:
        return None

    left = invertTree(root.right)
    right = invertTree(root.left)

    root.right = left
    root.left = right
    return root


node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(2)
node4 = TreeNode(3)
node5 = TreeNode(3)

node1.left = node2
node1.right = node3
node2.left = node4
node3.right = node5

invertTree(node1)
```

```python
# Balanced Binary Tree


class Solution:
    def isBalanced(self, root: TreeNode) -> bool:
        if root is None:
            return True

        leftHeight: int = self.__getHeight(root.right)
        rightHeight: int = self.__getHeight(root.left)

        if abs(leftHeight -rightHeight) > 1:
            return False

        return self.isBalanced(root.left) and self.isBalanced(root.right)

    def __getHeight(self, node: TreeNode) -> int:
        if node is None:
            return 0

        leftHeight: int = self.__getHeight(node.left)
        rightHeight: int = self.__getHeight(node.right)

        return max(leftHeight, rightHeight) + 1

node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(2)
node4 = TreeNode(3)
node5 = TreeNode(3)

node1.left = node2
node1.right = node3
node2.left = node4
node3.right = node5

Solution().isBalanced(node1)
```

```python
# Binary Tree Level Order Traversal
class Solution:
    def __init__(self) -> None:
        self.ans: list[list[int]] = list()

    def order(self, node: TreeNode, level: int) -> None:
        if len(self.ans) == level:
            self.ans.append(list())
        self.ans[level].append(node.val)

        if node.left is not None:
            self.order(node.left, level+1)

        if node.right is not None:
            self.order(node.right, level+1)

    def levelOrder(self, root: TreeNode) -> list[list[int]]:
        if root is None:
            return self.ans

        self.order(root, 0)
        return self.ans

tree = [3,9,20,None,None,15,7]

node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

Solution().levelOrder(node1)
```

```python
# Binary Tree ZigZag Level Order Traversal
from collections import deque

def zigzagLevelOrder(root: TreeNode) -> list[list[int]]:
    if root is None:
        return []

    results: list[list[int]] = list()
    node_queue = deque()
    node_queue.append(root)
    node_queue.append(None)

    level_list = deque()
    is_order_left = True

    while len(node_queue) > 0:
        curr_node: TreeNode = node_queue.popleft()
        if curr_node is not None:
            if is_order_left:
                level_list.append(curr_node.val)
            else:
                level_list.appendleft(curr_node.val)

            if curr_node.left is not None:
                node_queue.append(curr_node.left)
            if curr_node.right is not None:
                node_queue.append(curr_node.right)

        else:
            results.append(list(level_list))
            level_list = deque()

            if len(node_queue) > 0:
                node_queue.append(None)
            is_order_left = not is_order_left

    return results


    
node1 = TreeNode(3)
node2 = TreeNode(9)
node3 = TreeNode(20)
node4 = TreeNode(None)
node5 = TreeNode(None)
node6 = TreeNode(15)
node7 = TreeNode(7)

node1.left = node2
node1.right = node3
# node2.left = node4
# node2.right = node5
node3.left = node6
node3.right = node7

zigzagLevelOrder(node1)



```
