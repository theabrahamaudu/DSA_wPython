```python
## Factorial
## Method: Recursion
def findFactorial(n):
    if n == 0 or n == 1:
        return 1

    else:
        return n * findFactorial(n-1)


# Driver code
n = 7
result = findFactorial(n)
print(f"The factorial of {n} is", result)
```

```python
## Factorial
## Method: Iteration
def findFactorial(n):
    result = 1
    if n == 0 or n == 1:
        return result

    else:
        for i in range(2, n+1):
            result = result * i
    return result


# Driver code
n = 0
result = findFactorial(n)
print(f"The factorial of {n} is", result)
```

```python
## Fibonacci Series
# Method: Recursion
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)



# Driver Code 
n = 10
for i in range(n+1):
    print(f"{i} ->", fib(i))
```

```python
## Number of ways to reach upstairs
# Method: Recursion

def leadingFib(n):
    if n <= 1:
        return n
    else:
        return leadingFib(n-1) + leadingFib(n-2)

def ways(n):
    return leadingFib(n+1)



# Driver Code 
n = 30
for i in range(n+1):
    print(f"{i} ->", ways(i), "way(s)")
```

```python
## Number of ways to reach upstairs
# Method: Recursion, with caching

def leadingFib(n, memo = {}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    else:
        memo[n] = leadingFib(n-1, memo) + leadingFib(n-2, memo)
        return memo[n]

def ways(n):
    return leadingFib(n+1)



# Driver Code 
n = 101
for i in range(n+1):
    print(f"{i} ->", ways(i), "way(s)")
```
